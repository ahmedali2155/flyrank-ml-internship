# Predicting Content Click Opportunities Using Google Search Console Performance Data

## Abstract

This research investigates whether Google Search Console performance metrics can be used to identify content that is more likely to receive clicks. The analysis was conducted using the FlyRank ML Internship dataset containing search performance data from March 2026. A Random Forest Classifier was trained using search impressions and average search position and compared with a simple rule-based baseline under consistent evaluation settings. The model was further evaluated using grouped validation by client and a feature leakage audit to improve the reliability of the results. The findings provide observed and directional evidence that machine learning can support content optimization decisions, while emphasizing that the model should be used as a decision-support tool rather than a fully automated decision-making system.

---

# Research Question

**Can Google Search Console performance metrics be used to identify content with a higher likelihood of receiving clicks and support content optimization decisions?**

---

# Introduction / Problem Statement

Content teams often manage thousands of webpages but have limited time to review each one manually. Identifying which pages deserve optimization is an important but time-consuming task. Search performance metrics collected through Google Search Console provide useful signals that may help prioritize these decisions.

The objective of this project is to develop and evaluate a machine learning model that predicts whether a content item is likely to receive clicks using publicly safe search performance features. The project compares a Random Forest Classifier with a simple rule-based baseline using the same dataset and evaluation methodology.

To improve the credibility of the results, the project includes grouped validation by client, a feature leakage audit, and careful interpretation of the model's performance. These validation steps help reduce overly optimistic estimates and encourage responsible reporting of machine learning results.

The final outcome is a practical content action playbook that ranks optimization opportunities based on measured model outputs. The recommendations are intended to assist human reviewers in making better content decisions and should be interpreted as observed evidence on the evaluated dataset rather than proof of future performance.

# Methodology

## Objective

The objective of this project is to evaluate whether Google Search Console performance metrics can be used to identify content that is more likely to receive search clicks. The model is intended to support content optimization decisions rather than replace human judgment.

---

## Target Variable

The prediction target is a binary variable named **high_clicks**.

A content item is labeled as:

- **1** if `gsc_clicks > 0`
- **0** if `gsc_clicks = 0`

This converts the prediction task into a binary classification problem.

---

## Features

The model uses two search performance features:

- **gsc_impressions** – Number of times a page appeared in Google Search results.
- **gsc_sum_position** – Aggregate search position of the page.

These features were selected because they are available before prediction and do not directly encode the target variable.

---

## Baseline

Before training a machine learning model, a simple rule-based baseline was created using manually defined thresholds.

The Random Forest model was evaluated against this baseline using the same dataset so that the comparison would be fair and meaningful.

---

## Model Selection

A **Random Forest Classifier** was selected because it performs well on structured tabular datasets, handles non-linear relationships, requires minimal preprocessing, and is easy to implement using Scikit-learn.

The model was trained using default parameters with a fixed `random_state` to ensure reproducibility.

---

## Validation Strategy

Two validation approaches were used during the project.

### Week 5

- Random train/test split (80% training, 20% testing)

### Week 6

- Grouped validation using `client_hash_id`

The grouped validation prevents data from the same client appearing in both the training and testing sets, providing a more realistic estimate of model performance on unseen clients.

---

## Leakage Audit

A feature leakage audit was performed before evaluating the model.

The audit confirmed that:

- `gsc_clicks` was used only to create the target label.
- No future information was included.
- No label-derived features were used as model inputs.
- Client identifiers were used only for grouped validation and never as predictive features.

---

## Evaluation Metrics

Model performance was evaluated using:

- Accuracy
- Precision
- Recall
- F1-score
- Classification Report

The same evaluation metrics were used throughout the project to allow consistent comparison between the baseline model and the Random Forest model.

# Results

## Model Performance

The Random Forest Classifier was evaluated using the same dataset as the Week 4 baseline. Performance was measured using standard classification metrics and later re-evaluated using a grouped validation strategy to provide a more realistic estimate of generalization.

### Performance Summary

| Model | Validation Method | Accuracy |
|-------|-------------------|----------|
| Rule-Based Baseline | Random Split | Lower than Random Forest |
| Random Forest | Random Split | **94.13%** |
| Random Forest | Grouped Split | **95.63%** |

The grouped validation produced a similar level of performance while ensuring that data from the same client did not appear in both the training and testing sets. This validation approach provides a stronger estimate of how the model may perform on previously unseen clients.

---

## Classification Performance

The grouped validation produced the following evaluation metrics:

| Metric | Class 0 | Class 1 |
|---------|---------|---------|
| Precision | 0.98 | 0.43 |
| Recall | 0.98 | 0.41 |
| F1-score | 0.98 | 0.42 |

Overall Accuracy: **95.63%**

Although the overall accuracy is high, the model performs less effectively on the positive class (`high_clicks = 1`). This is likely due to class imbalance, where most content items belong to the negative class.

---

## Feature Importance

The Random Forest model identified the following feature importance values:

| Feature | Importance |
|---------|-----------:|
| gsc_sum_position | **0.5492** |
| gsc_impressions | **0.4508** |

The results indicate that search position contributed slightly more than search impressions when making predictions on the evaluated dataset.

The Random Forest model achieved higher measured performance than the simple rule-based baseline under the evaluation settings used in this project.

Grouped validation provided additional confidence that the evaluation was not overly optimistic by preventing data from the same client appearing in both the training and testing sets.

The feature importance analysis suggests that search position and search impressions both contribute meaningful predictive information. However, these observations should be interpreted as measured associations rather than evidence of causal relationships.

# Limitations & Honest Framing

## Limitations

This project was developed as a machine learning case study using the FlyRank ML Internship dataset. Although the model achieved strong measured performance on the evaluated dataset, several limitations should be considered.

### Limited Feature Set

Only two search performance features were used:

- `gsc_impressions`
- `gsc_sum_position`

Many other factors, such as content quality, search intent, page authority, metadata quality, and user behavior, may also influence whether a page receives clicks.

---

### Dataset Scope

The analysis was performed using approximately 100,000 records from the March 2026 dataset.

The findings therefore represent observations from this specific data release and should not be assumed to generalize to every website, client, or future time period.

---

### Class Imbalance

The target variable contains many more negative examples than positive ones.

As a result, overall accuracy is high, but the model is less effective at identifying pages that actually receive clicks. This can be observed in the lower precision and recall values for the positive class.

---

### Validation

Grouped validation using `client_hash_id` was performed to reduce optimistic evaluation caused by data from the same client appearing in both the training and testing sets.

While this provides a stronger estimate of model performance, additional validation using future time periods would further improve confidence in the results.

---

## Honest Claim Framing

The results presented in this paper should be interpreted using careful claim language.

This project demonstrates **observed** and **measured** relationships within the evaluated dataset. The findings provide **directional evidence** that search performance metrics can help support content optimization decisions.

The model is intended as a **decision-support tool** rather than an automated decision-making system. Human review remains necessary before implementing content recommendations.

No claim is made that the model explains Google's ranking algorithm or guarantees future search performance.
